```python
import json

from pypogo.game_master.game_master import GameMaster
```

```python
gm = GameMaster()

moves = {}
for move_name, move in gm._moves.items():
    moves[move_name] = move.to_dict()

with open('moves.json', 'w') as f:
    json.dump(moves, f)
```

```python
pokemon_data = {}

for pokemon_name, pokemon in gm._pokedex_pokemon.items():
    pokemon_data[pokemon_name] = pokemon.to_dict()

with open('pokemon.json', 'w') as f:
    json.dump(pokemon_data, f)
```
